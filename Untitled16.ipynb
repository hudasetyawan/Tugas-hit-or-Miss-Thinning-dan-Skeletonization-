```python
# Step 1: Install library jika perlu
!pip install scikit-image --quiet

# Step 2: Import library
import numpy as np
import matplotlib.pyplot as plt
from skimage.io import imread
from skimage.color import rgb2gray
from skimage.filters import threshold_otsu
from skimage.morphology import erosion, skeletonize, thin
from skimage.util import invert
from google.colab import files

# Step 3: Fungsi Hit-or-Miss
def hit_or_miss(image, se_foreground, se_background):
    image_complement = invert(image)  # Ambil komplemen dari gambar
    eroded_fg = erosion(image, se_foreground)  # Erosi gambar asli
    eroded_bg = erosion(image_complement, se_background)  # Erosi gambar komplemen
    return np.logical_and(eroded_fg, eroded_bg)  # Gabungkan dua hasil erosi

# Step 4: Upload gambar
uploaded = files.upload()
filename = next(iter(uploaded))

# Step 5: Baca gambar dan ubah ke biner
img = imread(filename)

# Jika gambar RGBA (memiliki channel alfa), kita hapus channel alfa
if img.shape[2] == 4:
    img = img[:, :, :3]  # Ambil hanya 3 channel (RGB)

gray = rgb2gray(img)  # Konversi ke grayscale
thresh = threshold_otsu(gray)  # Thresholding untuk binerisasi
binary = gray < thresh  # hitam = True

# Step 6: Structuring element foreground dan background untuk Hit-or-Miss
se_foreground = np.array([
    [0, 1, 0],
    [0, 1, 0],
    [0, 1, 0]
], dtype=bool)

se_background = np.array([
    [1, 0, 1],
    [1, 0, 1],
    [1, 0, 1]
], dtype=bool)

# Step 7: Hit-or-Miss
hitmiss_result = hit_or_miss(binary, se_foreground, se_background)

# Step 8: Thinning (Penipisan)
thinned_image = thin(binary)

# Step 9: Skeletonization (Pembuatan Skeletal)
skeleton_image = skeletonize(binary)

# Step 10: Visualisasi hasil
fig, axs = plt.subplots(2, 3, figsize=(18, 10))

axs[0, 0].imshow(binary, cmap='gray')
axs[0, 0].set_title("Gambar Biner")
axs[0, 0].axis('off')

axs[0, 1].imshow(hitmiss_result, cmap='gray')
axs[0, 1].set_title("Hit-or-Miss")
axs[0, 1].axis('off')

axs[0, 2].imshow(invert(binary), cmap='gray')
axs[0, 2].set_title("Komplemen")
axs[0, 2].axis('off')

axs[1, 0].imshow(thinned_image, cmap='gray')
axs[1, 0].set_title("Thinning")
axs[1, 0].axis('off')

axs[1, 1].imshow(skeleton_image, cmap='gray')
axs[1, 1].set_title("Skeletonization")
axs[1, 1].axis('off')

plt.tight_layout()
plt.show()
```
